1. Realice un programa traductor de frutas del español al inglés. Para ello, solicite el nombre de una fruta al usuario y muestre su traducción al inglés. Si la fruta no se encuentra en el diccionario, permítale al usuario agregarla con su respectiva traducción. El programa solo termina cuando el usuario ingresa una fruta que no está en el diccionario y no desea agregarla.

```python
frutas = {
    "piña": "pineapple",
    "banana": "banana",
    "naranja": "orange",
    "pera": "pear",
    "uva": "grape"
}

while True:
    fruta = input("Ingresa una fruta en español (o escriba 'salir' para terminar): ").lower()

    if fruta == "salir":
        print("¡Hasta luego!")
        break

    if fruta in frutas:
        print("Traducción:", frutas[fruta])
    else:
        print("La fruta no está en el diccionario.")
        agregar = input("¿Deseas agregarla? (si/no): ").lower()
        if agregar == "si":
            traduccion = input("Escribe su traducción al inglés: ")
            frutas[fruta] = traduccion
            print("Fruta agregada.")
        else:
            print("¡Hasta luego!")
            break
```

2. Desarrolla un traductor sencillo que utilice un diccionario donde las claves sean palabras en español y los valores sus traducciones en inglés. El usuario ingresará una palabra en español; si la palabra se encuentra en el diccionario, se mostrará su traducción. En caso contrario, se le solicitará al usuario que proporcione la traducción, la cual se agregará al diccionario. El proceso se repetirá hasta que el usuario decida finalizar la consulta.
SALIDA:

```python
diccionario = {
    "hola": "hello",
    "mundo": "world",
    "casa": "house"
}

print("Diccionario inicial:", diccionario)

while True:
    palabra = input("Ingresa una palabra en español (o 'salir' para terminar): ").lower()

    if palabra == "salir":
        print("Programa finalizado.")
        break

    if palabra in diccionario:
        print("Traducción:", diccionario[palabra])
    else:
        print("La palabra no está.")
        traduccion = input("Ingresa la traducción en inglés: ")
        diccionario[palabra] = traduccion
        print("Palabra agregada.")
```

3. Crea una aplicación que administre el inventario de una tienda usando un diccionario, donde cada clave representa el nombre de un producto y cada valor la cantidad en stock. La aplicación debe permitir consultar la cantidad de un producto, actualizar el stock de uno existente, agregar nuevos productos y eliminar aquellos que ya no estén disponibles. Al final se mostrará el inventario actualizado.

```python
inventario = {
    "pineapple": 50,
    "naranjas": 30,
    "peras": 20
}

print("Inventario inicial:", inventario)

print("Consulta: naranjas =", inventario["naranjas"])

inventario["peras"] = 25
print("Stock de peras actualizado a 25.")

inventario["bananas"] = 40
print("Producto 'bananas' agregado con 40 unidades.")

del inventario["naranjas"]
print("Producto 'naranjas' eliminado.")

print("Inventario final:", inventario)
```
